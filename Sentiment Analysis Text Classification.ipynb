# Load libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import classification_report, confusion_matrix
```

# Load dataset

```python
try:
    df = pd.read_csv('imdb_reviews.csv')
except Exception:
    # use small synthetic dataset
    df = pd.DataFrame({
    'review': ['I loved this movie', 'Terrible plot', 'It was okay', 'Best film ever', 'Not worth the time']*500,
    'sentiment': [1,0,1,1,0]*500
    })


X_train, X_test, y_train, y_test = train_test_split(df['review'], df['sentiment'], test_size=0.2, random_state=42)
```

# TF-IDF + Logistic Regression

```python
tfidf = TfidfVectorizer(max_features=10000, ngram_range=(1,2))
X_train_tfidf = tfidf.fit_transform(X_train)
X_test_tfidf = tfidf.transform(X_test)


clf = LogisticRegression(max_iter=1000)
clf.fit(X_train_tfidf, y_train)


preds = clf.predict(X_test_tfidf)
print(classification_report(y_test, preds))
```

# sample predictions

```python
samples = ['I really enjoyed this movie', 'Worst movie I have seen']
samples_tfidf = tfidf.transform(samples)
print('Predictions:', clf.predict(samples_tfidf))
```
